```python
import re, string
import numpy as np
import pandas as pd
import nltk
import gradio as gr
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

nltk.download('punkt', quiet=True)
nltk.download('stopwords', quiet=True)

def clean_text(text):
    if not isinstance(text, str): return ""
    text = text.lower().translate(str.maketrans('', '', string.punctuation))
    tokens = word_tokenize(text)
    stop_words = set(stopwords.words('english'))
    return " ".join([w for w in tokens if w not in stop_words])
```

```python
def get_hotel_data():
    return pd.DataFrame({
        "Question": [
            "What are the check-in and check-out timings?",
            "Is breakfast included in my stay?",
            "How can I book a table at the restaurant?",
            "Do you offer airport shuttle services?",
            "What is your cancellation policy?",
            "Is there free Wi-Fi and valet parking?",
            "Can I request an early check-in?",
            "Do you serve vegan or gluten-free food?"
        ],
        "Answer": [
            "Check-in starts at 2:00 PM and Check-out is at 11:00 AM. We offer luggage storage if you arrive early!",
            "Yes! A lavish continental breakfast is served from 7:00 AM to 10:30 AM in our main hall.",
            "You can reserve a table by calling our concierge or using the 'Dining' tab in our mobile app.",
            "We provide luxury private shuttles. Please share your flight details 24 hours before arrival.",
            "Cancellations are free up to 48 hours before check-in. After that, one night's charges apply.",
            "High-speed Wi-Fi is free for all guests. Valet parking is available at the main entrance 24/7.",
            "Early check-in is subject to room availability. We recommend calling us on the day of arrival.",
            "Our chefs specialize in dietary needs! We have a complete Vegan and Gluten-Free menu available."
        ]
    })

df = get_hotel_data()
df['Clean'] = df['Question'].apply(clean_text)
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(df['Clean'])
```

```python
def chat_engine(message, history):
    clean_query = clean_text(message)
    if not clean_query.strip():
        return "Welcome to **LuxeBot**. How can I assist you with your stay today?"

    query_vec = vectorizer.transform([clean_query])
    scores = cosine_similarity(query_vec, tfidf_matrix).flatten()
    idx = np.argmax(scores)

    # Background threshold set to 20% (No slider needed)
    if scores[idx] > 0.20:
        response = f"{df['Answer'].iloc[idx]}"
    else:
        response = "I'm sorry, I couldn't find exact details for that. Would you like to speak with our **Front Desk**?"

    # Suggestions add karna (Professional look ke liye)
    top_3 = np.argsort(scores)[::-1][:3]
    suggestions = [df['Question'].iloc[i] for i in top_3 if scores[i] > 0.1 and i != idx]

    if suggestions:
        response += "\n\n**You might also want to know:**\n" + "\n".join([f"• {s}" for s in suggestions])

    return response
```

```python
# Custom Luxury Theme
luxury_theme = gr.themes.Default(
    primary_hue="teal",
    secondary_hue="amber",
    neutral_hue="slate",
).set(
    body_background_fill="#0f172a", # Deep Dark Blue/Black
    block_background_fill="#1e293b", # Slate background
    block_title_text_color="#f59e0b", # Gold Title
    button_primary_background_fill="#f59e0b", # Gold Buttons
    button_primary_text_color="#000000"
)

with gr.Blocks(theme=luxury_theme) as demo:
    gr.Markdown("<h1 style='text-align: center; color: #f59e0b;'>✨ LuxeBot ✨</h1>")
    gr.Markdown("<p style='text-align: center; color: white;'>Your Virtual Concierge for a Seamless Stay</p>")

    # Simple Chat Interface - No Sidebar
    chatbot = gr.ChatInterface(
        fn=chat_engine,
        examples=["Breakfast timings?", "Parking info", "Can I cancel my booking?"]
    )

demo.launch(debug=True)
```
